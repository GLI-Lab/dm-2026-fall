## Option A (로컬): pixi + VSCode 또는 JupyterLab

### pixi를 쓰는 이유

conda나 mamba보다 환경 해결이 빠름. `pixi.lock`에 버전을 고정해서 학생마다 같은 패키지를 씀. conda 패키지와 pip 패키지를 한 워크플로에서 다룸.

### 사전 요구사항

아래 도구를 먼저 설치함.

> #### ❗ Windows 사용자
>
> 이 수업 환경은 **Linux**와 **macOS**만 지원함. Windows에서는 **WSL2**를 쓰고, Ubuntu 안에서 Linux 절차를 따름.
>
> 1. PowerShell을 관리자 권한으로 연 뒤 실행함:
> ```powershell
> wsl --install
> ```
> 2. 안내가 나오면 재시작함.
> 3. 시작 메뉴에서 **Ubuntu** 앱을 열어 초기 설정(사용자 이름·비밀번호)을 마침.
> 4. 이후 단계는 이 Ubuntu 터미널에서 실행함.
>
> WSL2 설치 안내: <https://learn.microsoft.com/en-us/windows/wsl/install>

**git.** 수업 저장소를 clone할 때 필요함.

- **Linux (WSL2 포함):** `sudo apt update && sudo apt install git`
- **macOS:** `git --version`으로 확인함. 없으면 macOS가 Xcode Command Line Tools 설치를 안내하거나, `xcode-select --install`을 실행함.

**pixi.** 빠른 크로스 플랫폼 패키지 관리자. `sudo`는 필요 없음. 사용자 계정 아래 `~/.pixi/`에 설치됨. 프로젝트 환경은 프로젝트 폴더의 `.pixi/envs/`에 둠. 설치 안내: <https://pixi.prefix.dev/latest/installation>.

```bash
# 시스템에 있는 쪽을 씀
curl -fsSL https://pixi.sh/install.sh | sh
# 또는
wget -qO- https://pixi.sh/install.sh | sh

# PATH 반영을 위해 셸을 다시 읽음
source ~/.bashrc     # zsh면 ~/.zshrc

# 설치 확인
pixi --version
# 예: pixi 0.63.2  (또는 그 이상)
```

**VSCode.** <https://code.visualstudio.com>에서 설치함. **Extensions** (`Ctrl+Shift+X` / `Cmd+Shift+X`)에서 Microsoft **Python** 확장을 설치함. Jupyter 노트북과 Python IntelliSense를 씀. Windows에서는 **WSL** 확장도 설치해서 VSCode가 WSL2 안 프로젝트를 열게 함.

### Step 1: 수업 저장소 clone

```bash
git clone -b students https://github.com/GLI-Lab/dm-2026-fall.git
cd dm-2026-fall
```

clone 후 저장소 구조는 아래와 같음.

```bash
dm-2026-fall/                  # 수업 저장소 루트
├── pixi.toml                  # 의존성 선언
├── pixi.lock                  # 고정된 버전
├── data/                      # 공유 데이터셋 (경로는 저장소 루트 기준)
├── assignments/               # 과제 스타터
└── exercises/                 # 실습 노트북
```

공유 데이터는 저장소 루트 `data/`에 둠. 실습은 `data.loader`를 import함. 로더가 그 경로에서 파일을 찾고, 없으면 받음. `lab01_setup.py`와 Lab 2 `helper.py`는 작업 폴더가 lab 폴더여도 저장소 루트를 `sys.path`에 넣음.

### Step 2: 패키지 설치

저장소 폴더에서 실행함.

```bash
pixi install
```

`pixi.lock`을 읽고 고정 버전으로 설치함. 목록은 `pixi.toml`에 있음.

```{.toml filename="pixi.toml"}
[dependencies]
python = "3.12.*"
numpy = "*"
scipy = "*"
pandas = "*"
scikit-learn = "*"
matplotlib = "*"
seaborn = "*"
nltk = "*"
networkx = ">=3.1"
mlxtend = "*"
gensim = ">=4.3.3"
jupyterlab = "*"
kagglehub = ">=1.0.2,<2"

[tasks]
lab = "jupyter lab --ip=0.0.0.0 --port=8888 --no-browser"
```

따로 설치할 필요 없음. `pixi install`이 처리함. 이 수업은 PyTorch나 GPU가 **필요 없음**.

> #### 📝 `pixi install` 동작
>
> `pixi.lock`이 있으면 (clone 후 보통 이 경우) solver 없이 고정 버전만 설치함. 빠름.
>
> lock 파일이 없으면 `pixi.toml`을 읽고 버전을 푼 뒤 `pixi.lock`을 쓰고 설치함. 한 번만 느림.
>
> `pixi.toml`이 바뀌면 (`pixi add <package>` 등) 다음 `pixi install` 또는 `pixi run`에서 lock이 갱신됨.

### Step 3: 실습 노트북 열기

JupyterLab 또는 VSCode로 염. 둘 다 pixi 환경을 씀.

**JupyterLab.** 저장소 루트에서 실행함.

```bash
pixi run lab
```

포트 8888에서 JupyterLab이 뜸. 브라우저는 자동으로 안 염. 터미널에 나온 URL을 브라우저에 붙임. 형태는 아래와 같음.

```bash
http://127.0.0.1:8888/lab?token=your_token_here
```

실습을 시작할 때마다 이 명령을 씀.

**VSCode.** 저장소 폴더를 연 뒤 pixi 커널을 고름.

1. VSCode에서 `.ipynb`를 염.
2. 노트북 오른쪽 위 **Select Kernel**을 누름.
3. **Python Environments**에서 저장소 안 `.pixi/envs/default/bin/python`을 고름.

> #### 📝 평소 작업
>
> 평소 실습은 `pixi run lab`이거나 VSCode에서 노트북을 여는 것으로 충분함. 셸을 오래 쓰려면 `pixi shell`로 환경을 켠 뒤 도구를 실행함. 끝나면 `exit`.

## Option B (클라우드): Google Colab

Google Colab은 로컬 설치 없이 브라우저에서 실습 노트북을 실행함. 각 노트북 상단 Colab 배지로 염.

![](https://colab.research.google.com/assets/colab-badge.svg)

Colab에는 `numpy`, `pandas`, `matplotlib`, `sklearn` 등이 이미 있음. 수업 전용 패키지(`networkx`, `nltk`, `mlxtend`, `gensim`, `kagglehub` 등)는 노트북 맨 위 setup 셀로 설치함.

```{.python}
# Google Colab에서만 이 셀을 실행함
!pip install <course-specific-package>
```

`data/` 아래 파일을 쓰는 노트북은 Colab에서 한 단계가 더 필요함. Colab은 로컬 디스크를 못 보므로, 세션 시작할 때 수업 저장소를 clone함.

```{.python}
# Google Colab에서만 이 셀을 실행함
!git clone -b students https://github.com/GLI-Lab/dm-2026-fall.git
```

이 셀은 노트북에 이미 있음. 나머지보다 먼저 실행함.

> #### ⚠️ Colab 제한
>
> Colab 환경은 **일시적**임. 설치한 패키지와 받은 파일은 세션이 끝나거나 끊기면 사라짐. 새 세션마다 setup 셀을 다시 실행함.
>
> 실습마다 같은 환경을 유지하려면 로컬 pixi (Option A)를 씀.
